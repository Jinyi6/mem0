```python
import json

ori = "membench824.json"
with open(ori, 'r', encoding='utf-8') as f:
    data = json.load(f)

subset = data[:100]

out_path = ori.replace('.json', '_100.json')
with open(out_path, 'w', encoding='utf-8') as f:
    json.dump(subset, f, ensure_ascii=False, indent=2)
```

```python
import json
# 下面这个json数据有10组。分成10份，每一个内包含1组数据。
locomo_path = "locomo10_failed.json"
with open(locomo_path, 'r', encoding='utf-8') as f:
    locomo_data = json.load(f)

# 将数据分成10份
locomo_chunks = [locomo_data[i:i + 1] for i in range(0, len(locomo_data), 1)]
for idx, chunk in enumerate(locomo_chunks):
    out_locomo_path = locomo_path.replace('.json', f'_{idx}.json')
    with open(out_locomo_path, 'w', encoding='utf-8') as f:
        json.dump(chunk, f, ensure_ascii=False, indent=2)
```
